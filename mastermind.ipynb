<summary>💡 Hilfe 6 — Musterlösung</summary>

Nur öffnen, wenn du wirklich nicht mehr weiterkommst — am meisten lernst du, wenn du selbst auf die Lösung kommst!

```python
import random

def erzeuge_code(laenge=4, farben=range(1, 7)):
    return random.choices(farben, k=laenge)

def lese_versuch(laenge=4, farben=range(1, 7)):
    while True:
        rohtext = input(f"Dein Versuch ({laenge} Ziffern zwischen {min(farben)} und {max(farben)}, z. B. 1 2 3 4): ")
        teile = rohtext.split()
        if len(teile) != laenge:
            print(f"Bitte genau {laenge} Werte eingeben.")
            continue
        try:
            versuch = [int(t) for t in teile]
        except ValueError:
            print("Bitte nur Ziffern eingeben.")
            continue
        if not all(v in farben for v in versuch):
            print(f"Bitte nur Werte zwischen {min(farben)} und {max(farben)} verwenden.")
            continue
        return versuch

def auswerten(code, versuch):
    rest_code = []
    rest_versuch = []
    schwarz = 0
    for c, v in zip(code, versuch):
        if c == v:
            schwarz += 1
        else:
            rest_code.append(c)
            rest_versuch.append(v)

    weiss = 0
    for v in rest_versuch:
        if v in rest_code:
            weiss += 1
            rest_code.remove(v)

    return schwarz, weiss

def spiele(max_versuche=10):
    code = erzeuge_code()
    print("Ich habe einen geheimen Code aus 4 Ziffern (1-6) gewählt. Viel Erfolg!")
    for versuch_nr in range(1, max_versuche + 1):
        versuch = lese_versuch()
        schwarz, weiss = auswerten(code, versuch)
        print(f"Versuch {versuch_nr}/{max_versuche}: {schwarz} schwarz, {weiss} weiss")
        if schwarz == len(code):
            print(f"🎉 Gewonnen nach {versuch_nr} Versuchen!")
            return
    print(f"Leider verloren. Der gesuchte Code war: {code}")

spiele()
```
