# Mastermind im Terminal

Implementiere das Spiel **Mastermind** als Python-Programm, das im Terminal gespielt wird.
Der Computer übernimmt die Rolle der Codemacherin, du errätst den Code.

## Kurz zur Erinnerung: Spielregeln

- Die Codemacherin (der Computer) legt einen **geheimen Code** aus 4 Farben fest. Zur Auswahl stehen `"rot"`, `"gelb"`, `"blau"` und `"gruen"`, Wiederholungen sind erlaubt. Ein Code ist also eine Liste wie `["rot", "gruen", "blau", "gelb"]` oder `["rot", "rot", "blau", "gelb"]`.
- Du gibst einen Versuch ein: vier Farben, jeweils einzeln in Kleinbuchstaben, z. B. `rot`, `gruen`, `blau`, `gelb`.
- Du erhältst als Rückmeldung:
  - **schwarze Stifte**: richtige Farbe an der richtigen Position
  - **weisse Stifte**: richtige Farbe, aber falsche Position
- Du hast maximal **10 Versuche**, um den Code zu erraten.

## Auftrag

Schreibe den Code für das Spiel in der Zelle unten. Mindestanforderungen:

1. Zufälliger geheimer Code aus 4 Farben (`"rot"`, `"gelb"`, `"blau"`, `"gruen"`), nicht sichtbar.
2. Eingabe eines Versuchs (4 Farben) im Terminal.
3. Berechnung und Anzeige von schwarzen/weissen Stiften nach jedem Versuch.
4. Maximal 10 Versuche.
5. Erkennung von Sieg/Niederlage; bei Niederlage wird der Code verraten.
6. Abfangen ungültiger Eingaben (Farbe nicht in der Liste der erlaubten Farben, z. B. Tippfehler oder Grossbuchstaben), ohne dass ein Versuch verloren geht.

## Gestufte Hilfen

Versuche die Aufgabe zuerst **selbstständig** zu lösen. Öffne die Hilfen erst, wenn du nicht weiterkommst — und immer nur **eine Stufe nach der anderen**. Klicke auf eine Hilfe, um sie aufzuklappen.

````python
# Dein Code für das Mastermind-Spiel

import random

farbliste = ["rot", "gelb", "blau", "gruen"]
farbcode = []
symbole = {"rot": "🔴", "blau": "🔵", "gruen": "🟢", "gelb": "🟡", "orange": "🟠", "lila": "🟣"}
print(" ".join(symbole[f] for f in ["rot", "blau", "gruen", "gelb"]))

def farbcode_generieren():
    # mach 4 mal
    for i in range(4):
        # Zufällige Position 
        zufallspos = random.randint(0,3)
        # Farbe an zufälliger Position
        farbe = farbliste[zufallspos]
        # Farbe Farbcode hinzufügen
        farbcode.append(farbe)
    return farbcode

def spieler_innenzug():
    vorschlag = []
    for i in range(4):
        farbe = input("Farbe kleingeschrieben eingeben: ")
        vorschlag.append(farbe)
    return vorschlag

def vergleichen(farbcode, vorschlag):
    if farbcode == vorschlag:
        print("Du hast gewonnen!")
    else:
        for position, farbe in enumerate(vorschlag):
            if farbe == farbcode[position]:
                print("Du hast eine Farbe mit richtiger Position gefunden!")
            elif farbe in farbcode:
                print("Du hast eine Farbe gefunden, aber mit falscher Position.")
            else:
                print("Du hast keine Farbe richtig.")
        
vergleichen(farbcode_generieren(), spieler_innenzug())
````

**1. Durchgang — schwarze Stifte:**
Vergleiche Code und Versuch Position für Position. Bei Übereinstimmung: schwarzer Stift. Merke dir, welche Farben dabei schon "verbraucht" wurden (z. B. in zwei separaten Listen der noch nicht zugeordneten Farben).

**2. Durchgang — weisse Stifte:**
Prüfe für die *übrig gebliebenen* (noch nicht als schwarz gezählten) Farben, ob eine Farbe aus dem Versuch noch irgendwo in den *übrig gebliebenen* Farben des Codes vorkommt. Falls ja: weisser Stift, und die Farbe wird aus der "Restliste" entfernt (damit sie nicht zweimal gezählt wird).

Ein möglicher Baustein dafür: Kopiere Code und Versuch in neue Listen. Entferne daraus zuerst alle exakten Treffer (schwarz). Gehe dann die Restliste des Versuchs durch und prüfe mit `in` und `.remove()`, ob die Farbe noch im Rest-Code vorkommt.

Ein Beispiel zum Testen: Bei Code `["rot", "rot", "blau", "gelb"]` und Versuch `["rot", "blau", "rot", "rot"]` gibt es 1 schwarzen und 2 weisse Stifte. Das dritte `"rot"` im Versuch bekommt keinen Stift mehr, weil beide `"rot"` im Code schon vergeben sind.
