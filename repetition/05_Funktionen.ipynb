# 05 · Funktionen

**Lernziele:** Du kannst eigene Funktionen mit Parametern definieren, Werte mit `return` zurückgeben und Funktionen aufrufen.

## Warum Funktionen?
Eine Funktion ist ein **benannter Code-Block**, den man beliebig oft wiederverwenden kann. Du kennst schon einige: `print()`, `len()`, `sum()`. Jetzt baust du eigene.

````python
def begruessen(name):
    print(f"Hallo {name}!")

begruessen("Anna")
begruessen("Luca")
````

## Aufbau
```python
def funktionsname(parameter1, parameter2):
    # eingerückter Block
    ergebnis = ...
    return ergebnis
```
- `def` leitet die Definition ein, am Ende der Kopfzeile steht ein **Doppelpunkt**.
- **Parameter** sind Platzhalter, beim Aufruf übergibst du **Argumente**.
- `return` gibt einen Wert **zurück** und beendet die Funktion.

### `print` ≠ `return`
`print` zeigt etwas auf dem Bildschirm an. `return` liefert einen Wert, mit dem man **weiterrechnen** kann.

````python
def verdopple(x):
    return 2 * x

y = verdopple(5) + 1     # funktioniert, weil verdopple einen Wert zurückgibt
print(y)
````

## Standardwerte
Parameter können einen Standardwert haben. Lässt du das Argument weg, wird dieser verwendet.

````python
def potenz(basis, exponent=2):
    return basis ** exponent

print(potenz(3))       # 9
print(potenz(2, 10))   # 1024
````

**Variablen in Funktionen sind lokal:** Was in einer Funktion definiert wird, existiert nur dort.

> ⚠️ Eine Funktion muss zuerst **definiert** (Zelle ausführen) und danach **aufgerufen** werden.

## Kurze Verständnisfragen
Überlege zuerst selbst, dann klappe die Antwort auf.

<details>
<summary>❓ Frage 1: Was gibt eine Funktion ohne `return` zurück?</summary>

`None`, einen speziellen «Nichts»-Wert. Wer das Ergebnis weiterverwenden will, braucht `return`.

</details>

<details>
<summary>❓ Frage 2: Was passiert mit Code, der in der Funktion **nach** einem `return` steht?</summary>

Er wird nicht mehr ausgeführt, denn `return` beendet die Funktion sofort.

</details>

<details>
<summary>❓ Frage 3: Was ist der Unterschied zwischen Parameter und Argument?</summary>

Parameter stehen in der Definition (`def f(x)`), Argumente sind die konkreten Werte beim Aufruf (`f(5)`).

</details>
