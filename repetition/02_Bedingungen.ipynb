**Lernziele:** Du kannst Werte vergleichen, Wahrheitswerte kombinieren und mit `if` / `elif` / `else` Entscheidungen im Programm treffen.

## Vergleiche
Ein Vergleich ergibt einen `bool`: `True` oder `False`.

| Operator | Bedeutung |
|---|---|
| `==` | gleich |
| `!=` | ungleich |
| `<` `>` `<=` `>=` | kleiner, grösser, kleiner gleich, grösser gleich |

````python
print(5 == 5, 5 != 5, 3 < 7, 7 <= 7)
print("a" == "A")
````

## `if`, `elif`, `else`
Nach der Bedingung steht ein **Doppelpunkt**; der zugehörige Block ist **um 4 Leerzeichen eingerückt**. Die Einrückung gehört in Python zur Syntax.

````python
temperatur = 22

if temperatur > 25:
    print("Heiss")
elif temperatur > 15:
    print("Angenehm")
else:
    print("Kühl")
````

Es wird **nur der erste** passende Zweig ausgeführt. Deshalb ist die Reihenfolge der Bedingungen wichtig.

## `and`, `or`, `not`
- `a and b`: beides muss stimmen
- `a or b`: mindestens eines muss stimmen
- `not a`: kehrt um

````python
alter = 17
hat_ticket = True
print(alter >= 16 and hat_ticket)
print(alter < 12 or alter > 65)
print(not hat_ticket)
print(10 < alter < 20)      # Python erlaubt verkettete Vergleiche
````

<details>
<summary>❓ Frage 1: Warum liefert `if x = 5:` einen Fehler?</summary>

Für Vergleiche braucht es `==`. Ein einzelnes `=` ist eine Zuweisung und an dieser Stelle nicht erlaubt.

</details>

<details>
<summary>❓ Frage 2: Was gibt dieser Code aus? `x = 8` / `if x > 5: print('A')` / `elif x > 7: print('B')`</summary>

Nur `A`. Sobald ein Zweig zutrifft, werden die übrigen `elif`/`else` übersprungen.

</details>

<details>
<summary>❓ Frage 3: Was ergibt `not (True and False)`?</summary>

`True`. Zuerst `True and False` = `False`, dann `not False` = `True`.
