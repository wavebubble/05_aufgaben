## Strings sind Folgen von Zeichen
Wie bei Listen gibt es Index, Slicing, `len()` und Schleifen. Strings sind aber **unveränderlich**: Du kannst ein Zeichen nicht überschreiben, nur einen neuen String bauen.

````python
wort = "Informatik"
print(wort[0], wort[-1], len(wort))
print(wort[2:5])         # "for"
print(wort[::-1])        # rückwärts
for zeichen in wort[:3]:
    print(zeichen)
````

## Wichtige Methoden
Methoden werden mit einem Punkt aufgerufen. Sie liefern einen **neuen** String zurück.

| Methode | Wirkung |
|---|---|
| `s.upper()`, `s.lower()` | Gross-/Kleinbuchstaben |
| `s.strip()` | Leerzeichen am Rand entfernen |
| `s.replace("a", "b")` | ersetzen |
| `s.count("a")` | Vorkommen zählen |
| `s.find("a")` | Position des ersten Treffers (`-1`, wenn keiner) |
| `s.split()` | in eine Liste von Wörtern zerlegen |
| `"-".join(liste)` | Liste zu String zusammenfügen |
| `s.isalpha()`, `s.isdigit()` | nur Buchstaben / nur Ziffern? |
| `"x" in s` | Teilstring enthalten? |

````python
satz = "  Python macht Spass  "
print(satz.strip().upper())
woerter = satz.split()
print(woerter)
print("-".join(woerter))
print("macht" in satz)
````

## Zeichen und Codes
Jedes Zeichen hat eine Nummer (Unicode). `ord("a")` liefert sie (97), `chr(97)` wandelt zurück. Da die Kleinbuchstaben aufeinanderfolgend nummeriert sind, kann man damit rechnen, was für die Verschlüsselung in Aufgabe 6.6 nützlich wird.

````python
print(ord("a"), ord("b"), chr(100))
print(ord("d") - ord("a"))    # Abstand im Alphabet
````

<details>
<summary>💡 Musterlösung anzeigen</summary>

```python
def caesar(text, verschiebung):
    ergebnis = ""
    for z in text:
        if "a" <= z <= "z":
            pos = (ord(z) - ord("a") + verschiebung) % 26
            ergebnis += chr(ord("a") + pos)
        else:
            ergebnis += z
    return ergebnis
```
