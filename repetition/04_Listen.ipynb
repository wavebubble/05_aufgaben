# 04 · Listen

**Lernziele:** Du kannst Listen erstellen, auf Elemente zugreifen, sie verändern, über sie iterieren und Teillisten (Slices) bilden.

## Grundlagen
Eine Liste speichert mehrere Werte in einer Reihenfolge. Die Zählung beginnt bei **0**. Negative Indizes zählen von hinten.

````python
farben = ["rot", "grün", "blau", "gelb"]
print(farben[0])      # erstes Element
print(farben[-1])     # letztes Element
print(len(farben))    # Länge
farben[1] = "violett" # Element ändern
print(farben)
````

## Wichtige Operationen

| Code | Wirkung |
|---|---|
| `liste.append(x)` | `x` hinten anfügen |
| `liste.remove(x)` | erstes `x` entfernen |
| `liste.pop()` | letztes Element entfernen und zurückgeben |
| `x in liste` | ist `x` enthalten? (`True`/`False`) |
| `sum(liste)`, `min(liste)`, `max(liste)` | Summe, Minimum, Maximum |
| `sorted(liste)` | sortierte Kopie |

## Über Listen iterieren

````python
zahlen = [3, 8, 1, 6]
for z in zahlen:                 # direkt über die Elemente
    print(z * 2)

for i in range(len(zahlen)):     # über die Positionen
    print(i, zahlen[i])
````

## Slicing
`liste[start:ende]` liefert die Elemente ab `start` **bis vor** `ende`. `liste[::-1]` ergibt die umgekehrte Liste.

````python
z = [10, 11, 12, 13, 14, 15]
print(z[1:4])    # [11, 12, 13]
print(z[:3])     # [10, 11, 12]
print(z[3:])     # [13, 14, 15]
print(z[::-1])   # umgekehrt
````

## Kurze Verständnisfragen
Überlege zuerst selbst, dann klappe die Antwort auf.

<details>
<summary>❓ Frage 1: Welches Element ist `liste[len(liste)]`?</summary>

Keines: das ergibt einen `IndexError`. Das letzte Element hat den Index `len(liste) - 1` (oder `-1`).

</details>

<details>
<summary>❓ Frage 2: Was ergibt `[1, 2, 3, 4, 5][1:3]`?</summary>

`[2, 3]` (Index 1 bis vor Index 3).

</details>

<details>
<summary>❓ Frage 3: Was ist der Unterschied zwischen `sorted(l)` und `l.sort()`?</summary>

`sorted(l)` liefert eine **neue** sortierte Liste und lässt `l` unverändert. `l.sort()` sortiert `l` selbst und gibt nichts zurück.

</details>
