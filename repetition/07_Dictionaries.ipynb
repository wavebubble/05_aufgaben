# 07 · Dictionaries

**Lernziele:** Du kannst Daten als Schlüssel-Wert-Paare speichern, Einträge lesen, ändern und löschen, über Dictionaries iterieren und Häufigkeiten zählen.

## Idee
Eine Liste findest du über die **Position**, ein Dictionary über einen **Schlüssel** (wie ein Wörterbuch: Wort → Bedeutung). Schlüssel sind eindeutig.

````python
person = {"name": "Anna", "alter": 17, "ort": "Zürich"}
print(person["name"])
person["alter"] = 18            # ändern
person["hobby"] = "Klettern"    # neu hinzufügen
del person["ort"]               # löschen
print(person)
print(len(person))
````

## Abfragen und Durchlaufen
- `"name" in person`: gibt es den Schlüssel?
- `person.get("ort", "unbekannt")`: Wert oder Ersatzwert (kein Fehler, wenn der Schlüssel fehlt)
- `person["fehlt"]` ergibt einen `KeyError`
- `.keys()`, `.values()`, `.items()`: Schlüssel, Werte, Paare

````python
noten = {"Mathe": 5, "Deutsch": 4.5, "Informatik": 6}

print("Mathe" in noten, "Bio" in noten)
print(noten.get("Bio", "keine Note"))

for fach in noten:                    # iteriert über die Schlüssel
    print(fach, "->", noten[fach])

for fach, note in noten.items():      # Schlüssel und Wert gleichzeitig
    print(f"{fach}: {note}")

print(sum(noten.values()) / len(noten))
````

    h = {}
    for z in text:
        if z in h:
            h[z] += 1
        else:
            h[z] = 1
    return h

## Kurze Verständnisfragen
Überlege zuerst selbst, dann klappe die Antwort auf.

<details>
<summary>❓ Frage 1: Was ist der Unterschied zwischen `d["x"]` und `d.get("x")`, wenn `"x"` fehlt?</summary>

`d["x"]` ergibt einen `KeyError`. `d.get("x")` gibt `None` zurück (oder den angegebenen Ersatzwert, z. B. `d.get("x", 0)`).

</details>

<details>
<summary>❓ Frage 2: Was liefert `for k in d:` bei einem Dictionary?</summary>

Die **Schlüssel**. Für Werte nimmst du `d.values()`, für beides `d.items()`.

</details>

<details>
<summary>❓ Frage 3: Darf ein Schlüssel zweimal vorkommen?</summary>

Nein. Wird ein Schlüssel nochmals zugewiesen, überschreibt der neue Wert den alten.

</details>
