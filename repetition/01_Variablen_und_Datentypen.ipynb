# 01 · Variablen und Datentypen

**Lernziele:** Du kannst Werte in Variablen speichern, mit Zahlen rechnen, Datentypen unterscheiden und umwandeln und Text mit f-Strings ausgeben.

## Variablen
Eine Variable ist ein **Name für einen Wert**. Mit `=` weist du ihr einen Wert zu.

````python
alter = 17
name = "Anna"
print(name, "ist", alter, "Jahre alt")
alter = alter + 1      # Variable überschreiben
print(alter)
````

Das `=` bedeutet **«speichere in»**, nicht «ist gleich». Die Zeile `alter = alter + 1` heisst: «Nimm den alten Wert, rechne +1 und speichere ihn wieder in `alter`».

**Regeln für Namen:** Buchstaben, Ziffern und `_`; nicht mit einer Ziffer beginnen; keine Leerzeichen; Gross-/Kleinschreibung zählt (`Alter` ≠ `alter`). Nimm sprechende Namen (`flaeche`, nicht `x7`).

## Datentypen

| Typ | Beispiel | Bedeutung |
|---|---|---|
| `int` | `42` | ganze Zahl |
| `float` | `3.14` | Kommazahl (mit **Punkt**!) |
| `str` | `"Hallo"` | Text (String) |
| `bool` | `True`, `False` | Wahrheitswert |

Mit `type()` fragst du den Typ ab.

````python
print(type(42), type(3.14), type("Hallo"), type(True))
````

## Rechnen

| Operator | Bedeutung | Beispiel | Ergebnis |
|---|---|---|---|
| `+` `-` `*` | wie gewohnt | `3 * 4` | `12` |
| `/` | Division (immer `float`) | `7 / 2` | `3.5` |
| `//` | ganzzahlige Division | `7 // 2` | `3` |
| `%` | Rest (Modulo) | `7 % 2` | `1` |
| `**` | Potenz | `2 ** 10` | `1024` |

Es gilt Punkt-vor-Strich; mit Klammern steuerst du die Reihenfolge.

````python
print(7 / 2, 7 // 2, 7 % 2, 2 ** 10)
print(2 + 3 * 4, (2 + 3) * 4)
````

## Umwandeln und Ausgeben
- `int("5")`, `float("2.5")`, `str(7)` wandeln Typen um.
- **f-Strings** setzen Werte direkt in Text ein: `f"... {variable} ..."`.
- `input("Frage")` liest Text ein. Das Ergebnis ist **immer ein `str`**, bei Zahlen musst du also umwandeln.

````python
preis = 12.5
anzahl = 3
print(f"{anzahl} Stück kosten {preis * anzahl} Franken")

# input() liefert immer Text:
# alter = int(input("Wie alt bist du? "))
````
