## So arbeitest du mit den Notebooks

- Ein Notebook besteht aus **Zellen**. Es gibt **Text-Zellen** (Markdown) und **Code-Zellen**.
- Eine Zelle führst du mit **Shift + Enter** aus. Die Ausgabe erscheint direkt darunter.
- Zellen werden **von oben nach unten** ausgeführt. Variablen aus früheren Zellen bleiben erhalten.
- Läuft etwas komisch? Menü **Kernel → Restart Kernel and Run All Cells**. Das räumt alles auf.
- **Selbsttest-Zellen** prüfen deine Lösung. Erscheint `✓ … ist richtig!`, hast du es geschafft.
- **Musterlösungen** sind versteckt. Versuche es zuerst selbst, mindestens 5 Minuten lang!

**Schwierigkeit:** ★☆☆ einfach · ★★☆ mittel · ★★★ knifflig

## Fehlermeldungen lesen

Fehler sind normal, auch Profis machen sie ständig. Wichtig ist, die Meldung zu lesen.
**Schau immer zuerst auf die letzte Zeile der Meldung.**

| Fehler | Bedeutung | Typische Ursache |
|---|---|---|
| `SyntaxError` | Python versteht die Zeile nicht | Doppelpunkt, Klammer oder Anführungszeichen fehlt |
| `NameError` | Name unbekannt | Tippfehler oder Variable noch nicht definiert (Zelle vergessen auszuführen?) |
| `TypeError` | Falscher Datentyp | `"5" + 3`: Text und Zahl lassen sich nicht addieren |
| `IndentationError` | Falsch eingerückt | Nach `:` muss der Block 4 Leerzeichen eingerückt sein |
| `ZeroDivisionError` | Division durch 0 | |
| `IndexError` | Listenposition gibt es nicht | Index zu gross |
