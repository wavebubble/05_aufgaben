# 03 · Schleifen

**Lernziele:** Du kannst Anweisungen mit `for` und `while` wiederholen, `range()` benutzen und Werte in einer Schleife aufsummieren.

## `for` mit `range()`
`range(n)` liefert die Zahlen `0, 1, …, n-1`. Variationen:

| Aufruf | Zahlen |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(2, 6)` | 2, 3, 4, 5 |
| `range(0, 10, 3)` | 0, 3, 6, 9 (Schrittweite 3) |

Das Ende ist **nie** dabei.

````python
for i in range(5):
    print("Durchgang", i)
````

## Akkumulator-Muster
Eine häufige Aufgabe: Werte aufsummieren. Vor der Schleife wird eine Variable auf 0 gesetzt, in der Schleife wird sie erhöht.

````python
summe = 0
for zahl in range(1, 6):
    summe = summe + zahl      # kurz: summe += zahl
print(summe)                  # 1+2+3+4+5
````

## `while`
Die Schleife läuft, **solange** die Bedingung wahr ist. Verwende sie, wenn du **nicht im Voraus** weisst, wie oft wiederholt wird. Achtung: Ändere in der Schleife etwas, das die Bedingung irgendwann falsch macht, sonst läuft sie endlos! (Notfall: Kernel → Interrupt.)

Mit `break` verlässt du eine Schleife vorzeitig.

````python
n = 1
while n < 100:
    n = n * 2
print(n)

for i in range(10):
    if i == 3:
        break
    print(i)
````
